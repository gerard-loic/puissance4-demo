```python

import pickle

l = [1, 2, 3, 4]
class SimpleClass():
	def __init__(self, nom):
		self.nom = nom
		
a = SimpleClass(nom="Moi")

#Ecriture
with open("test.pkl", "wb") as f:
	pickle.dump(l, f)
	pickle.dump(a, f)
	
#Chargement
with open("test.pkl", "rb") as f:
	for i in range(2):
		obj = pickle.load(f)
		print(obj)
```
