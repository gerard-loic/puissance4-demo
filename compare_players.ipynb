```python
import sys
sys.path.append('game/')
from tictactoe import TicTacToe
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
```

```python
players = [
    'random_player',
    'minimax_player',
    'dynamicprogramming_player',
    'montecarlo_player',
    'temporaldifference_player'
]

results_matrix = np.array([
    [0.308, -0.807, -0.813, -0.862, -0.201],  # random vs tous
    [0.994, 0, 0, 0, 1.0],  # minimax vs tous
    [0.995, 0, 0, 0, 1.0],  # dynamicprogramming vs tous
    [0.99, 0, 0, 0.655, 1.0],  # montecarlo vs tous
    [0.709, -1, -1, -1, 1],  # temporaldifference vs tous
])

nb_games = 1000
```

```python
for player1 in players:
    for player2 in players:
        print(f"Evaluating agent {player1} against {player2} on {nb_games} games")

        total_score = 0
        for _ in tqdm(range(nb_games), 'Games : '):
            game = TicTacToe(player_one=player1, player_two=player2, display_board=False)
            total_score += game.playGame(returnWinner=True)

        efficiency_score = total_score/nb_games

        results_matrix[players.index(player1)][players.index(player2)] = efficiency_score
        print(f"Efficiency : {efficiency_score}")

```

```python
# Création de la heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    results_matrix,
    annot=True,          # Afficher les valeurs
    fmt='.2f',           # Format des nombres (2 décimales)
    cmap='RdYlGn',       # Palette de couleurs (Rouge-Jaune-Vert)
    xticklabels=players,
    yticklabels=players,
    cbar_kws={'label': 'Taux de victoire'},
    vmin=-1,              # Minimum de l'échelle
    vmax=1,              # Maximum de l'échelle
    linewidths=0.5,      # Lignes de séparation
    square=True          # Cellules carrées
)

plt.title('Matrice de performance - Taux de victoire (Ligne vs Colonne)', 
          fontsize=14, fontweight='bold', pad=20)
plt.xlabel('Adversaire', fontsize=12, fontweight='bold')
plt.ylabel('Agent', fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Calcul des statistiques globales
print("\n=== Statistiques globales ===")
print(f"{'Agent':<30} {'Winrate moyen':<15} {'Meilleur contre':<25} {'Pire contre'}")
print("-" * 100)

for i, player in enumerate(players):
    # Exclure les matchs contre soi-même
    winrates = [results_matrix[i, j] for j in range(len(players)) if i != j]
    avg_winrate = np.mean(winrates)
    
    # Meilleur et pire adversaire
    results_vs_others = [(players[j], results_matrix[i, j]) 
                         for j in range(len(players)) if i != j]
    best_matchup = max(results_vs_others, key=lambda x: x[1])
    worst_matchup = min(results_vs_others, key=lambda x: x[1])
    
    print(f"{player:<30} {avg_winrate:.2%}{'':>9} "
          f"{best_matchup[0]:<25} {worst_matchup[0]}")
```
